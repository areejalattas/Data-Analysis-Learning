```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('seaborn-v0_8')
```

```python
df=pd.read_csv('/kaggle/input/competitions/house-prices-advanced-regression-techniques/train.csv')
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
plt.scatter(df['OverallQual'],df['SalePrice'],alpha=0.5)

plt.title('Quality vs SalePrice')
plt.xlabel('Quality')
plt.ylabel('SalePrice')

# Result => As the quality increases, the price increases (Positive correlation)
```

```python
plt.scatter(df['YearBuilt'], df['SalePrice'], c=df['OverallQual'],s=df['OverallCond']*10 )
cbar = plt.colorbar()
cbar.set_label('Quality')

plt.title('YearBuilt vs SalePrice')
plt.xlabel('YearBuilt')
plt.ylabel('SalePrice')

# Conclusion => Newer houses (higher YearBuilt) tend to have higher quality and higher prices (Positive correlation)
```
