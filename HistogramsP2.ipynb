```python
import matplotlib.pyplot as plt
#matplotlib.style.use('seaborn-v0_8')
%matplotlib inline
```

```python
import pandas as pd 
df=pd.read_csv(r"C:\Users\MSIi\Documents\Data engineer\DataCSV\500_Person_Gender_Height_Weight_Index.csv")
df.head()
```

```python
heights = df['Height']
bins = [i for i in range(140, 201, 5)]

plt.hist(heights, bins=bins, edgecolor='black')
plt.title('Height Distribution')
plt.xlabel('Heights')
plt.ylabel('Frequency')
plt.show()
```
