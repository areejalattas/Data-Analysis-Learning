Histograms

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
%matplotlib inline
plt.style.use('seaborn-v0_8')
```

```python
ages = [10, 30, 45, 30, 31, 46, 12, 16, 32, 23, 28, 29]
bins = [10, 20, 30, 40, 50]

# edgecolor => set bar border color
# bins => divide data into intervals (ranges) along the x-axis

plt.hist(ages, edgecolor='black', bins=bins)

# Add a line for a specific value (e.g., mean)
mean = np.mean(ages)                                    # mean = 27.67
plt.axvline(mean, color='red', linewidth=2)             # axvline => Axis Vertical Line

plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```
