```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('seaborn-v0_8')
```

link for matplotlib color maps ==>https://matplotlib.org/stable/gallery/color/colormap_reference.html

```python
x = [5, 7, 8, 5, 6, 7, 9, 2, 3, 4, 4, 4, 2, 6, 3, 6, 8, 6, 4, 1]
y = [7, 4, 3, 9, 1, 3, 2, 5, 2, 4, 8, 7, 1, 6, 4, 9, 7, 7, 5, 1]
colors = [9, 10, 1, 7, 5, 7, 3, 5, 6, 7, 1, 10, 8, 1, 9, 2, 1, 6, 7, 5]
sizes = [209, 486, 381, 255, 191, 315, 185, 228, 174, 538, 239, 394, 399, 153, 273, 293, 436, 501, 397, 539]

# color => fixed circle color
# c => array of colors per circle
# cmap => color map theme (e.g., 'autumn')
# s => circle size
# edgecolor => border color
# linewidth => border thickness
# alpha => transparency (0.0 to 1.0)

plt.scatter(x, y, c=colors, s=sizes, edgecolor='black', linewidth=2, alpha=0.5, cmap='autumn')

cbar = plt.colorbar()               # display color scale bar
cbar.set_label('Color Bar')         # set label for the color bar

plt.title('Data')
plt.xlabel('X Data')
plt.ylabel('Y Data')
plt.show()
```
