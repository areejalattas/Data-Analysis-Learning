Pie Charts

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.style.use('ggplot')
```

```python
data = [30, 20, 40, 10]
labels = ['Python', 'JS', 'Html', 'C++']
colors = ['red', 'black', 'orange', 'green']      # Colors for the slices (names or hex codes)
explode = [0, 0, 0, 0.2]                          # Pulls slices outward to highlight them (0 = none)

plt.title('Programmer')
plt.pie(
    data,
    labels=labels,                                # Labels/titles for each pie slice
    colors=colors,                                # Applies the slice colors list
    explode=explode,                              # Offsets specific slices from the center
    shadow=True,                                  # Adds a drop shadow beneath the pie chart
    startangle=90,                                # Starting angle in degrees for the first slice
    wedgeprops={'edgecolor': 'black'},            # Wedge border/styling properties
    autopct='%1.1f%%',                            # Displays formatted percentage labels on slices
    rotatelabels=True                             # Rotates label text along each slice's angle
)
```
